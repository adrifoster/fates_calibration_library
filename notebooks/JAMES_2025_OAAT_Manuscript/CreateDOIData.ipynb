```python
description_clm = "Global monthly output from a parameter perturbation experiment with CLM."
description_fates = "Global monthly output from a parameter perturbation experiment with CLM-FATES (with FATES parameters)."
description_fatesclm = "Global monthly output from a parameter perturbation experiment with CLM-FATES (with CLM parameters)."

mini_oaat2 = " In this experiment, stomatal conductance parameters were aligned with CLM's"
mini_oaat3 = " In this experiment, stomatal conductance parameters and water stress appliction were aligned with CLM's"

descriptions = [description_clm,
                description_fates,
                description_fatesclm,
                description_fates + mini_oaat2,
                description_fates + mini_oaat3,
                description_fatesclm + mini_oaat2,
                description_fatesclm + mini_oaat3]
```

```python
files = climatology_files

for i, file in enumerate(files):

    # update attributes of variables
    dat = xr.open_dataset(os.path.join(in_dir, file))

    # dat.coords["lat"].attrs["long_name"] = "latitude"
    # dat.coords["biome"].attrs["biome_names_map"] = biome_names_json
    # dat.coords["biome"].attrs["long_name"] = "Whittaker biome index (see biome_names_map attribute for names)"
    # dat['summation_var'] = dat.summation_var.assign_attrs(description='mean or interannual variance')
    
    data_vars = dat.data_vars
    for var in data_vars:
        dat[var] = dat[var].assign_attrs(
            units=climatology_units[var], description=long_names[var]
        )
    
    # add metadata
    creation_date = datetime.now().strftime("%Y-%m-%d %H:%M:%S UTC")
    
    dat.attrs['owner'] = 'Adrianna Foster'
    dat.attrs['contact'] = 'afoster@ucar.edu'
    dat.attrs['date'] = creation_date
    dat.attrs['description'] = descriptions[i]
    dat.to_netcdf(os.path.join(hist_dir, file))
```
